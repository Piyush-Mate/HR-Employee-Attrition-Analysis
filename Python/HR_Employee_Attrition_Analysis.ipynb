```python
summary = pd.DataFrame({
    "Metric": [
        "Overall Attrition",
        "Attrition - Overtime",
        "Attrition - No Overtime",
        "Attrition - Age 18-25",
        "Attrition 0-5 Years Experience",
        "Attrition - Single + Overtime",
        "Attrition - Low Job Involvement + Overtime"
    ],
    "Rate": [
        (df["Attrition"] == "Yes").mean() * 100,
        df.loc[df["OverTime"] == "Yes", "Attrition"].eq("Yes").mean() * 100,
        df.loc[df["OverTime"] == "No", "Attrition"].eq("Yes").mean() * 100,
        df.loc[
            df["Age"].between(18, 25),
            "Attrition"
        ].eq("Yes").mean() * 100,
        df.loc[
            df["TotalWorkingYears"].between(0, 5),
            "Attrition"
        ].eq("Yes").mean() * 100,
        df.loc[
            (df["OverTime"] == "Yes") &
            (df["MaritalStatus"] == "Single"),
            "Attrition"
        ].eq("Yes").mean() * 100,
        df.loc[(df["OverTime"] == "Yes") &
               (df["JobInvolvement"] == 1),
               "Attrition"
        ].eq("Yes").mean() * 100,
    ]
})


summary["Rate"] = summary["Rate"].round(2)

summary
```
